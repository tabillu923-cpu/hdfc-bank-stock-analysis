```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

hdfc = pd.read_csv("HDFCBANK.csv")

hdfc["Date"] = pd.to_datetime(
    hdfc["Date"], dayfirst=True, errors="coerce"
)

hdfc = hdfc.dropna(subset=["Date", "Close"])

hdfc["Daily_Return"] = hdfc["Close"].pct_change() * 100
hdfc["Year"] = hdfc["Date"].dt.year

print(hdfc.shape)
print(hdfc[["Date", "Close", "Daily_Return", "Year"]].head())
```

```python
annual = hdfc.groupby("Year").agg(
    Annual_Return=(
        "Daily_Return",
        lambda x: ((1 + x.dropna() / 100).prod() - 1) * 100
    ),
    Daily_Volatility=("Daily_Return", "std")
)

annual = annual.dropna()

annual.round(2).head(10)
```

```python
best_year = annual["Annual_Return"].idxmax()
worst_year = annual["Annual_Return"].idxmin()
highest_risk_year = annual["Daily_Volatility"].idxmax()

print("Best Year:", best_year)
print("Best Annual Return:", round(annual.loc[best_year, "Annual_Return"], 2), "%")

print("\nWorst Year:", worst_year)
print("Worst Annual Return:", round(annual.loc[worst_year, "Annual_Return"], 2), "%")

print("\nHighest Daily Volatility Year:", highest_risk_year)
print("Daily Volatility:", round(annual.loc[highest_risk_year, "Daily_Volatility"], 2), "%")
```

```python
# Calculate returns using adjusted closing prices
hdfc["Daily_Return"] = hdfc["Adj Close"].pct_change() * 100

# Ensure the Year column exists
hdfc["Year"] = hdfc["Date"].dt.year

# Calculate annual return and daily volatility
annual = hdfc.groupby("Year").agg(
    Annual_Return=(
        "Daily_Return",
        lambda x: ((1 + x.dropna() / 100).prod() - 1) * 100
    ),
    Daily_Volatility=("Daily_Return", "std")
)

# Convert daily volatility to annualized volatility
annual["Annualized_Volatility"] = (
    annual["Daily_Volatility"] * np.sqrt(252)
)

annual.round(2).head(10)
```

```python
best_year = annual["Annual_Return"].idxmax()
worst_year = annual["Annual_Return"].idxmin()
highest_risk_year = annual["Annualized_Volatility"].idxmax()
lowest_risk_year = annual["Annualized_Volatility"].idxmin()

print("Best Year:", best_year)
print("Best Return:", round(annual.loc[best_year, "Annual_Return"], 2), "%")

print("\nWorst Year:", worst_year)
print("Worst Return:", round(annual.loc[worst_year, "Annual_Return"], 2), "%")

print("\nHighest Risk Year:", highest_risk_year)
print("Annualized Volatility:", round(annual.loc[highest_risk_year, "Annualized_Volatility"], 2), "%")

print("\nLowest Risk Year:", lowest_risk_year)
print("Annualized Volatility:", round(annual.loc[lowest_risk_year, "Annualized_Volatility"], 2), "%")
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(14, 7))

plt.scatter(
    annual["Annualized_Volatility"],
    annual["Annual_Return"]
)

for year in annual.index:
    plt.annotate(
        str(year),
        (
            annual.loc[year, "Annualized_Volatility"],
            annual.loc[year, "Annual_Return"]
        ),
        fontsize=8
    )

plt.axhline(y=0, linestyle="--")
plt.xlabel("Annualized Volatility (%)")
plt.ylabel("Annual Return (%)")
plt.title("HDFC Bank: Annual Return vs Annualized Volatility")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
